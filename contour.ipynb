```python
import cv2
import numpy as np
```

```python
defect_mask = cv2.imread(r"C:\vs code\PCB_circuit\results\01_missing_hole_1_20\defect_mask_1_01.jpg", cv2.IMREAD_GRAYSCALE)

contours, hierarchy = cv2.findContours(defect_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
```

```python
test_img = cv2.imread(r"C:\vs code\PCB_circuit\PCB_DATASET\PCB_DATASET\images\Missing_hole\01_missing_hole_01.jpg")
```

```python
for cnt in contours:
    area = cv2.contourArea(cnt)
    if area > 100:  
        x, y, w, h = cv2.boundingRect(cnt)  
        cv2.rectangle(test_img, (x, y), (x+w, y+h), (0, 255, 255), 2)
```

```python
cv2.imwrite("defect_with_contours.jpg", test_img)

print(f"Number of defects detected: {len(contours)}")
```
