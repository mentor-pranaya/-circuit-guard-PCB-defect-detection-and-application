```python
import cv2
import numpy as np

import os
```

```python

output_dir = r"C:\vs code\PCB_circuit\ROI_Output\14_missing_hole_1_20"

#04_20  05_10   06_10  07_10  08_10 09_10 10_05 11_10 12_10
for i in range(1,11):
    defect_mask_path = fr"C:\vs code\PCB_circuit\results\11_missing_hole_1_10\defect_mask_11_{i:02d}.jpg"
    test_img_path = fr"C:\vs code\PCB_circuit\PCB_DATASET\PCB_DATASET\images\Missing_hole\11_missing_hole_{i:02d}.jpg"
    defect_mask = cv2.imread(defect_mask_path, cv2.IMREAD_GRAYSCALE)


    test_img = cv2.imread(test_img_path)

    if defect_mask is None:
        raise FileNotFoundError(f"Defect mask not found: {defect_mask_path}")
    if test_img is None:
        raise FileNotFoundError(f"Test image not found: {test_img_path}")


    contours, _ = cv2.findContours(defect_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)


    contour_img = test_img.copy()
    cv2.drawContours(contour_img, contours, -1, (0, 200, 255), 2)  # red contours

    contour_img_rgb = cv2.cvtColor(contour_img, cv2.COLOR_BGR2RGB)

    os.makedirs(output_dir, exist_ok=True)
    roi_count=0
    for j, cnt in enumerate(contours):
        area = cv2.contourArea(cnt)
        if area > 50:  # filter small noise
            x, y, w, h = cv2.boundingRect(cnt)
            roi = test_img[y:y+h, x:x+w]  # Crop ROI
            roi_path = os.path.join(output_dir, f"11_roi_{i:02d}_{roi_count}.jpg")
            cv2.imwrite(roi_path, roi)
            print(f"✅ Saved ROI: {roi_path}")
            roi_count+=1
    cv2.imwrite(fr"C:\vs code\PCB_circuit\ROI_Output\14_missing_hole_1_20\defect_with_contours_{i:02d}.jpg", contour_img_rgb)
    print(f"Number of defects detected: {len(contours)}")
    print(f"No of ROI:{roi_count}")
```
