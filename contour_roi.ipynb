```python
import cv2
import numpy as np
```

```python
roi_results=r"C:\vs code\PCB_circuit\resultsROI_Output\\12_roi_output_1_10"
import os
os.makedirs(roi_results,exist_ok=True)
```

```python
for i in range(1,11):
    defect_mask = cv2.imread(fr"C:\vs code\PCB_circuit\results\12_missing_hole_1_10\defect_mask_12_{i:02d}.jpg", cv2.IMREAD_GRAYSCALE)

    contours, hierarchy = cv2.findContours(defect_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    test_img = cv2.imread(fr"C:\vs code\PCB_circuit\PCB_DATASET\PCB_DATASET\images\Missing_hole\12_missing_hole_{i:02d}.jpg")
    roi_count=0
    for cnt in contours:
        area = cv2.contourArea(cnt)
        if area > 100:  
            x, y, w, h = cv2.boundingRect(cnt)  
            cv2.rectangle(test_img, (x, y), (x+w, y+h), (0, 255, 255), 2)
            roi=test_img[y:y+h,x:x+w]
            #roi_filename=os.path.join(roi_results,f"1_roi_1_{i:02d}_{roi_count}.jpg")
            #success = cv2.imwrite(roi_filename, roi)
            #if success:
             #   print(f"Saved: {roi_filename}")
            ##   print(f"Failed to save: {roi_filename}")
            cv2.imwrite(fr"C:\vs code\PCB_circuit\ROI_Output\12_roi_output_1_10\12_roi_{i:02d}_{roi_count}.jpg",roi)
            roi_count+=1
    cv2.imwrite(fr"C:\vs code\PCB_circuit\ROI_Output\12_roi_output_1_10\defect_with_contours_{i:02d}.jpg", test_img)

    print(f"Number of defects detected: {len(contours)}")
    print(f"No of ROI:{roi_count}")
```
