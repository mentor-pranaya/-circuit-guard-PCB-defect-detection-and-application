```python
import cv2
import numpy as np
import os

#01_missing_hole_20  04_20  05_10   06_10  07_10  08_10 09_10 10_05 11_10 12_10
template_path = r"C:\vs code\PCB_circuit\PCB_DATASET\PCB_DATASET\PCB_USED\01.JPG"
save_path = r"C:\vs code\PCB_circuit\results\01_missing_hole_1_20"
os.makedirs(save_path, exist_ok=True)
for i in range(1,21):
    images_path =fr"C:\vs code\PCB_circuit\PCB_DATASET\PCB_DATASET\images\Missing_hole\01_missing_hole_{i:02d}.jpg"
    test = cv2.imread(images_path, cv2.IMREAD_GRAYSCALE)
    if test is None:
        print(f"Skipping {images_path}, not found.")
        continue
    test_img = cv2.imread(images_path, cv2.IMREAD_GRAYSCALE)
    template_img = cv2.imread(template_path, cv2.IMREAD_GRAYSCALE)
    #test_img = cv2.resize(test_img, (640, 640))
    #template_img = cv2.resize(template_img, (640, 640))
    diff = cv2.absdiff(template_img, test_img)
    blurred = cv2.GaussianBlur(diff, (5, 5), 0)
    _, defect_mask = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    test_color = cv2.cvtColor(test_img, cv2.COLOR_GRAY2BGR)
    test_color[defect_mask > 0] = [0, 0, 255] 
    cv2.imwrite(os.path.join(save_path, f"difference_1_{i:02d}.jpg"), diff)
    cv2.imwrite(os.path.join(save_path, f"defect_mask_1_{i:02d}.jpg"), defect_mask)
    cv2.imwrite(os.path.join(save_path, f"defect_highlighted_1_{i:02d}.jpg"), test_color)

    print("✅ Processing done! Results saved in:", save_path)
```
